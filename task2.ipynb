```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv("/student_performance_clustering_dataset.csv")
df
```

```python
X = df[["Study_Hours_Per_Day","Attendance_Percentage","Assignment_Score"]]
X.head()
```

```python
kmeans =KMeans(n_clusters=5,random_state=42,n_init=10)
kmeans.fit(X)
df["Cluster"] = kmeans.labels_
df.head()
```

```python
plt.scatter(
    X["Study_Hours_Per_Day"],
    X["Assignment_Score"],
    c=df["Cluster"]
)

plt.xlabel("Study Hours")
plt.ylabel("Assignment Score")
plt.title("Student Performance using K-Means")
plt.show()
```
